```python
import re
from datasets import load_dataset
```

```python
TARGET = 300_000
```

```python
wiki = load_dataset("wikimedia/wikipedia", "20231101.uk", split="train", streaming=True)
sentences = []
```

```python
for article in wiki:
    text = article["text"].lower()                        

    for sent in re.split(r"[.!?]+", text):                
        sent = re.sub(r"[^а-яіїєґ'’ʼ\- ]", " ", sent)     
        words = [w.strip("-'’ʼ") for w in sent.split()]   
        words = [w for w in words if len(w) >= 3]         

        if len(words) >= 5:                               
            sentences.append(words)

    if len(sentences) >= TARGET:
        break

sentences = sentences[:TARGET]
print(len(sentences))
print(sentences[:2])
```
